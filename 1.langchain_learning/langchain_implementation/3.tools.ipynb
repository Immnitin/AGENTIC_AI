## Tools
Models that requests functions to fetch data  or do any tasks.
it is pairing of schema and function to execute

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ['OPENAI_API_KEY']=os.getenv('OPENAI_API_KEY')
os.environ['GROQ_API_KEY']=os.getenv('GROQ_API_KEY')
os.environ['GOOGLE_API_KEY']=os.getenv('GOOGLE_API_KEY')
```

### Method-1

```python
from langchain.tools import tool
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

model=init_chat_model("groq:qwen/qwen3-32b")


@tool
def get_weather(city:str)->str:
    '''this tool is used to get the weather of the given location'''
    return f"the weather of {city} is sunny"

model_with_tools=model.bind_tools([get_weather])

response=model_with_tools.invoke("get the weather of hydebad")

for tool_call in response.tool_calls:
    print(tool_call["name"], end="||")
    print(tool_call["args"])

```

Method-2

```python
from langchain.agents import create_agent
model=create_agent(
    model="groq:qwen/qwen3-32b",
    tools=[get_weather],
    system_prompt={'you are a helping assistant'}
)
```

Tool Execution Cycle
1) Step 1: for the input the llm checks how many tools are needed.
2) Step 2: Then the input message is run againist the tools and append the responses
3) Step 3: Then print the final response from the response
