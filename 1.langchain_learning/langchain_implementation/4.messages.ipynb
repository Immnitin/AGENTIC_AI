- Messages Prompt:
Messages are the fundemental units of context. It consists of role, content, metadata
- Role: type of message
- Content: it is the actual content that message consists of.
- Metadata: it consists of other information about the message calls, tool calls
Langchain gives the standard format of the message that helps maintain consistency

Messages are four types AI-Message, Human-Message, System-Message, Tool-Message

- <b>System Message</b> : is like the instruction given to the agent on how to answer or behave and giving it basic context 
- <b>Human Message</b> : it consists of actual content of the message , of what the user wants to ask to the agent
- <b>AI Message</b> : it consists of the response of the AI. 
- <b>Tool Message</b> : Consists of the details regarding the tool calls like the tool name, arguements passed and all

```python
from langchain.messages import AIMessage,HumanMessage,SystemMessage,ToolMessage

messages=[
    SystemMessage("You are an expert Python developer"),
    HumanMessage("What are closures in python and how are they used "),
]

tool_result="weather 123"
messages.append(ToolMessage(tool_result,tool_call_id=12))

model.invoke(messages)
```
