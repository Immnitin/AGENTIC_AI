# Skills in DeepAgents
    On demand expertise with progressive disclosure

Skills are specialized capability modules that are loaded by agents only when they are relevant to the user request

```python
import os
from dotenv import load_dotenv
load_dotenv()
GOOGLE_API_KEY=os.getenv('GOOGLE_API_KEY')
```

```python
from deepagents import create_deep_agent
from deepagents.backends import StateBackend
from deepagents.backends.utils import create_file_data
from langgraph.checkpoint.memory import MemorySaver
from pathlib import Path
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite",api_key=GOOGLE_API_KEY)

skill_dirs=['aws','python','langgraph','report_writer']

skill_files={f'skills/{name}/skill.md':create_file_data(Path(f'skills/{name}/skill.md').read_text(encoding='utf-8')) for name in skill_dirs}

skill_files
```

```python
agent=create_deep_agent(
    model=model,
    memory=['skills/'],
    checkpointer=MemorySaver()
)
config={'configurable':{'thread_id':'demo1'}}
```

```python
response=agent.invoke({'messages':[{'role':'user','content':'What all skills you have in your memory and when do you use them'}],'files':skill_files},config=config)
response
```

```python
response=agent.invoke({'messages':[{'role':'user','content':'How do i build a Lnaggraph with conditional routing '}],'files':skill_files},config=config)
```
