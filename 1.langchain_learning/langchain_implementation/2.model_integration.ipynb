```python
from langchain.chat_models import init_chat_model

model=init_chat_model("gpt-4.1")
response=model.invoke("Hello How are you")
response        
```

Groq chat models Integration

```python
model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("Hello how are you")
response.content
```

```python
from langchain_groq import ChatGroq

model=ChatGroq(model="qwen/qwen3-32b")
```

## Streaming is streaming the output content while the content is being generated, it improves the user experience

```python
for chunks in model.stream("Why do parrots have feather"):
    print(chunks.text, end="", flush=True)
```

Batch is processing of multiple requests as a batch, it significantly improves the performance and the cost

```python
responses=model.batch(["why parrots have feathers","how parrots can speak","how do parrots travel","how do parrots see"])
for response in responses:
    print(response.content)
```
